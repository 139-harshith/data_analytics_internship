```python
import pandas as pd

# Load dataset
df = pd.read_csv("stock_data_big.csv")

# Show data
print(df.head())

# Check dataset size (Big Data proof)
print("Total Rows:", len(df))

# Average closing price per company
avg_close = df.groupby("Company")["Close"].mean()
print("\nAverage Closing Price:\n", avg_close)

# Highest stock price
max_price = df["High"].max()
print("\nHighest Stock Price:", max_price)

# Total trading volume per company
total_volume = df.groupby("Company")["Volume"].sum()
print("\nTotal Volume:\n", total_volume)

# Sort companies by performance
print("\nTop Companies by Avg Close:")
print(avg_close.sort_values(ascending=False))
```

```python
import pandas as pd
import random
from datetime import datetime, timedelta

# Create data
rows = 100000  # 1 lakh rows

companies = ["RELIANCE", "TCS", "INFY", "HDFC", "ICICI"]

data = []

start_date = datetime(2024, 1, 1)

for i in range(rows):
    date = start_date + timedelta(days=random.randint(0, 30))
    company = random.choice(companies)
    
    open_price = random.randint(1000, 4000)
    high = open_price + random.randint(0, 200)
    low = open_price - random.randint(0, 200)
    close = random.randint(low, high)
    volume = random.randint(50000, 200000)
    
    data.append([date, company, open_price, high, low, close, volume])

df = pd.DataFrame(data, columns=["Date","Company","Open","High","Low","Close","Volume"])

df.to_csv("stock_data_big.csv", index=False)

print("Dataset created successfully!")
```

```python
import matplotlib.pyplot as plt

avg_close.plot(kind='bar', title="Average Closing Price by Company")
plt.xlabel("Company")
plt.ylabel("Price")
plt.show()
```
